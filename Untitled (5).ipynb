# PubMedQA Fine-Tuning & Evaluation
This notebook presents an end-to-end comparison between a baseline large language model
and a LoRA fine-tuned version on the PubMedQA dataset.

The task consists of answering biomedical research questions with a single label:
yes, no, or maybe, based on the provided scientific context.

The workflow includes:
- Preparing and formatting the PubMedQA dataset
- Running inference with the base (pretrained) model
- Fine-tuning the model using LoRA (parameter-efficient fine-tuning)
- Evaluating both models on a held-out test set
- Comparing performance using Accuracy and Macro-F1 metrics

The objective is to measure the impact of LoRA fine-tuning compared to the baseline model
under the same inference and evaluation conditions.
Ce notebook présente l’ensemble du processus de préparation des données, d’entraînement et d’évaluation d’un modèle de langage appliqué à la tâche de question-réponse binaire/ternaire sur le dataset PubMedQA. Il compare les performances d’un modèle de base pré-entraîné et de sa version fine-tunée avec LoRA, en analysant à la fois l’évolution de la perte pendant l’entraînement et les métriques finales de classification sur un jeu de test dédié.

## Dataset Preparation

In this section, we load the PubMedQA (pqa_labeled) dataset and prepare it for training and evaluation.

Each example is converted into an instruction-style prompt composed of:
- a QUESTION
- a CONTEXT
- an explicit instruction to answer with one word only: yes, no, or maybe

The dataset is then split into:
- 80% training
- 10% validation
- 10% test

These splits are used later for fine-tuning and for a fair comparison between
the baseline model and the LoRA fine-tuned model.


```python
def convert(example):
    question = example["question"]
    context = example["context"]
    label = example["final_decision"]

    prompt = (
        f"QUESTION: {question}\n"
        f"CONTEXT:\n{context}\n\n"
        "Answer with one word only: yes, no, or maybe."
    )

    return {
        "input": prompt,
        "output": label.lower()
    }

processed = raw.map(convert)
processed
```

```python
split = processed["train"].train_test_split(
    test_size=0.2,
    seed=42
)

tmp = split["test"].train_test_split(
    test_size=0.5,
    seed=42
)

dataset = {
    "train": split["train"],
    "validation": tmp["train"],
    "test": tmp["test"],
}

```

## Load Base Model & Tokenizer
This section loads the pretrained Mistral-7B Instruct model along with its corresponding tokenizer.
The tokenizer is configured to use the end-of-sequence token as the padding token to ensure compatibility during batching and generation.

The base language model is loaded in inference-ready mode, using automatic device placement (device_map="auto") and automatic precision selection (torch_dtype="auto").

```python
model_name = "mistralai/Mistral-7B-Instruct-v0.2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token

base_model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    torch_dtype="auto"
)
```

## Tokenization (Train + Validation)


```python
def tokenize(batch):
    texts = batch["input"]
    tokens = tokenizer(
        texts,
        truncation=True,
        padding="max_length",
        max_length=512,
    )
    tokens["labels"] = tokens["input_ids"].copy()
    return tokens
tokenized_train = dataset_raw["train"].map(
    tokenize,
    batched=True,
    remove_columns=["input", "output"],
)

tokenized_val = dataset_raw["validation"].map(
    tokenize,
    batched=True,
    remove_columns=["input", "output"],
)


tokenized = dataset_raw.map(
    tokenize,
    batched=True,
    remove_columns=["input", "output"]
)
```

## Fine-Tuning (LoRA)

```python
training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=3,
    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,

    logging_steps=10,              
    evaluation_strategy="steps",   
    eval_steps=10,
    save_steps=10,

    report_to="none",
    fp16=True,
)

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_val,
    data_collator=data_collator,
)

trainer.train()
```

This cell defines the function used to generate the model’s prediction from a given prompt. It tokenizes the input, runs deterministic text generation (without sampling) to produce a short answer, then decodes only the newly generated tokens to extract the final model output as clean, lowercase text without special tokens.Cette cellule définit la fonction utilisée pour générer la prédiction du modèle à partir d’un prompt donné. Elle tokenise l’entrée, exécute une génération déterministe (sans échantillonnage) pour produire une réponse courte, puis décode uniquement les nouveaux tokens générés afin d’extraire la réponse finale du modèle sous forme de texte normalisé (minuscules, sans tokens spéciaux).

```python
def generate_label(model, tokenizer, prompt):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=3,   # ← PAS 1
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    decoded = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[-1]:],
        skip_special_tokens=True
    ).strip().lower()

    return decoded
```

```python
def normalize_prediction(text):
    if not text:
        return "unknown"

    text = text.lower().strip()

    if text.startswith("yes"):
        return "yes"
    if text.startswith("no"):
        return "no"
    if text.startswith("maybe"):
        return "maybe"

    return "unknown"

```

This cell defines the evaluation loop used to assess a model on a dataset. It iterates over each example, builds the prompt, generates a prediction using the model, normalizes the predicted text into a valid label (yes, no, or maybe), and collects both true and predicted labels, optionally limiting the number of evaluated samples.


```python
def evaluate_model(model, tokenizer, dataset, max_samples=None):
    y_true, y_pred = [], []

    for i, ex in enumerate(dataset):
        if max_samples and i >= max_samples:
            break

        prompt = ex["input"]
        true_label = ex["output"].lower().strip()

        if not prompt:
            continue

        pred_raw = generate_label(model, tokenizer, prompt)
        pred_label = normalize_prediction(pred_raw)

        y_true.append(true_label)
        y_pred.append(pred_label)

    return y_true, y_pred
```

This section evaluates and compares the baseline model and the LoRA fine-tuned model on the test set by computing standard classification metrics.
```python
from sklearn.metrics import accuracy_score, f1_score, classification_report

LABEL_MAP = {"yes": 0, "no": 1, "maybe": 2}
ALL_LABELS = [0, 1, 2]
ALL_NAMES = ["yes", "no", "maybe"]

def compute_metrics_clean(y_true, y_pred, name):
    y_true_ids = []
    y_pred_ids = []

    for t, p in zip(y_true, y_pred):
        if t in LABEL_MAP and p in LABEL_MAP:
            y_true_ids.append(LABEL_MAP[t])
            y_pred_ids.append(LABEL_MAP[p])

    print(f"\n=== {name} ===")


    if len(y_true_ids) == 0:
        print("Aucun sample exploitable")
        return {
            "Model": name,
            "Accuracy": 0.0,
            "Macro-F1": 0.0,
        }

    acc = accuracy_score(y_true_ids, y_pred_ids)
    f1 = f1_score(y_true_ids, y_pred_ids, average="macro", zero_division=0)

    print(classification_report(
        y_true_ids,
        y_pred_ids,
        labels=ALL_LABELS,
        target_names=ALL_NAMES,
        zero_division=0
    ))

    print("Accuracy:", acc)
    print("Macro-F1:", f1)

    return {
        "Model": name,
        "Accuracy": acc,
        "Macro-F1": f1,
    }
```

```python
from sklearn.metrics import classification_report, f1_score

label_map = {"yes": 0, "no": 1, "maybe": 2}

def compute_metrics(y_true, y_pred, name="Model"):
    y_true_ids, y_pred_ids = [], []

    for t, p in zip(y_true, y_pred):
        if t in label_map and p in label_map:
            y_true_ids.append(label_map[t])
            y_pred_ids.append(label_map[p])

    print(f"\n=== {name} ===")
    print("Samples valides:", len(y_true_ids))

    if len(y_true_ids) == 0:
        print("Aucun sample exploitable")
        return None

    print(classification_report(
        y_true_ids,
        y_pred_ids,
        target_names=["yes", "no", "maybe"],
        zero_division=0
    ))

    macro_f1 = f1_score(y_true_ids, y_pred_ids, average="macro")
    print("Macro-F1:", macro_f1)

    accuracy = sum(
        t == p for t, p in zip(y_true_ids, y_pred_ids)
    ) / len(y_true_ids)

    print("Accuracy:", accuracy)

    return accuracy, macro_f1
```
