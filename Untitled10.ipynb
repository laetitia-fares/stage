## Multimodal Task Encoder
Energon fournit au TaskEncoder des échantillons déjà décodés : du texte sous forme de chaînes, des images sous forme d’objets, des vidéos sous forme de décodeurs paresseux, et des métadonnées sous forme de dictionnaires. Le TaskEncoder décide comment transformer tout cela en entrées exploitables par le modèle. Il peut tokeniser le texte, transformer les images, extraire quelques frames d’une vidéo plutôt que la décoder entièrement, et attacher des informations sur l’origine du sample.

```python
class MultimodalTaskEncoder(DefaultTaskEncoder):
    """
    Production-grade task encoder:
    - handles text, image, audio, video
    - supports auxiliary media
    - supports packing & batching
    """

    def __init__(self):
        super().__init__()
        self.av_decoder = AVDecoder(
            decode_video=True,
            decode_audio=False,
            sample_frames=8
        )

    def encode_sample(self, sample: Sample) -> Dict[str, Any]:
        data = sample.data

        result = {}

        # TEXT
        if "txt" in data:
            result["text"] = data["txt"].strip()

        # IMAGE
        if "jpg" in data or "png" in data:
            result["image"] = data.get("jpg") or data.get("png")

        # VIDEO
        if "mp4" in data:
            video = self.av_decoder(data["mp4"])
            result["video_frames"] = video.video

        # AUXILIARY METADATA
        if "json" in data:
            result["meta"] = data["json"]

        # SUBFLAVORS (dataset origin tracking)
        result["source"] = sample.subflavor

        return result
```

## Packing Logic (Sequence Efficiency)
Le packing consiste à regrouper plusieurs petits échantillons en un seul échantillon plus grand, afin d’utiliser efficacement la longueur de contexte des modèles modernes. Energon effectue ce packing avant le batching, et il impose que cette opération soit strictement déterministe. C’est indispensable pour que la reprise après interruption fonctionne correctement. Cette cellule explique comment plusieurs morceaux de texte ou plusieurs éléments multimodaux peuvent être combinés sans casser la reproductibilité de l’entraînement.

```python
def pack_samples(samples: List[Dict[str, Any]]) -> Dict[str, Any]:
    packed = {
        "text": [],
        "images": [],
        "videos": [],
        "sources": []
    }

    for s in samples:
        if "text" in s:
            packed["text"].append(s["text"])
        if "image" in s:
            packed["images"].append(s["image"])
        if "video_frames" in s:
            packed["videos"].append(s["video_frames"])
        packed["sources"].append(s["source"])

    return packed
```

## Training Dataset Creation

À partir du metadataset, des règles de shuffle, des limites de packing, de la configuration des workers et du TaskEncoder, Energon construit un objet dataset qui sait déjà comment produire des batches cohérents. Contrairement à PyTorch classique, le batching et le shuffling ne sont pas délégués à un DataLoader externe : ils font partie intégrante du dataset Energon, ce qui garantit un comportement identique sur toutes les machines.

```python
dataset = get_train_dataset(
    path="metadataset.yaml",
    batch_size=8,
    shuffle_buffer_size=10_000,
    max_samples_per_sequence=4,
    worker_config=WorkerConfig(
        num_workers=8,
        worker_id=0,
        world_size=1
    ),
    task_encoder=MultimodalTaskEncoder(),
)
```
