## Imports

```python
import torch
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling,
)
from peft import LoraConfig, get_peft_model
```

## Load a generative dataset

```python
raw = load_dataset("squad_v2")

raw
```

## Convert examples to instruction-style prompts (generative answers)

```python
def convert(example):
    question = example["question"]
    context = example["context"]

    # Some SQuAD v2 questions have no answer
    if len(example["answers"]["text"]) > 0:
        answer = example["answers"]["text"][0]
    else:
        answer = "No answer available."

    prompt = (
        f"QUESTION: {question}\n"
        f"CONTEXT:\n{context}\n\n"
        "Answer the question in a complete sentence."
    )

    return {
        "input": prompt,
        "output": answer
    }

processed = raw.map(convert)
processed
```

## Train / Validation / Test split

```python
split = processed["train"].train_test_split(
    test_size=0.2,
    seed=42
)

tmp = split["test"].train_test_split(
    test_size=0.5,
    seed=42
)

dataset = {
    "train": split["train"],
    "validation": tmp["train"],
    "test": tmp["test"],
}

for k in dataset:
    print(k, len(dataset[k]))

```

## Save to JSONL and reload

```python
dataset["train"].to_json("/workspace/squad/train.jsonl", orient="records", lines=True)
dataset["validation"].to_json("/workspace/squad/val.jsonl", orient="records", lines=True)
dataset["test"].to_json("/workspace/squad/test.jsonl", orient="records", lines=True)

data_files = {
    "train": "/workspace/squad/train.jsonl",
    "validation": "/workspace/squad/val.jsonl",
    "test": "/workspace/squad/test.jsonl",
}

dataset_raw = load_dataset("json", data_files=data_files)
dataset_raw
```

## Load base model and tokenizer

```python
model_name = "mistralai/Mistral-7B-Instruct-v0.2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token

base_model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    torch_dtype="auto"
)
```

## Apply LoRA

```python
lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
)

model = get_peft_model(base_model, lora_config)
model.print_trainable_parameters()
```

## Tokenization (generative labels)

```python
def tokenize(batch):
    tokens = tokenizer(
        batch["input"],
        truncation=True,
        padding="max_length",
        max_length=512,
    )
    tokens["labels"] = tokenizer(
        batch["output"],
        truncation=True,
        padding="max_length",
        max_length=128,
    )["input_ids"]

    return tokens

tokenized_train = dataset_raw["train"].map(
    tokenize,
    batched=True,
    remove_columns=["input", "output"],
)

tokenized_val = dataset_raw["validation"].map(
    tokenize,
    batched=True,
    remove_columns=["input", "output"],
)
```

## Training setup

```python
training_args = TrainingArguments(
    output_dir="./results_fast",
    num_train_epochs=3,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    logging_steps=20,
    evaluation_strategy="steps",
    eval_steps=200,
    save_steps=200,

    fp16=True,
    report_to="none",
)


data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_val,
    data_collator=data_collator,
)

trainer.train()
```

## Plot training / validation loss

```python
train_loss, eval_loss, steps = [], [], []

for log in trainer.state.log_history:
    if "loss" in log and "step" in log:
        train_loss.append(log["loss"])
        steps.append(log["step"])
    if "eval_loss" in log:
        eval_loss.append(log["eval_loss"])

plt.figure(figsize=(8, 5))
plt.plot(steps[:len(train_loss)], train_loss, label="train loss")
plt.plot(steps[:len(eval_loss)], eval_loss, label="eval loss")
plt.xlabel("Training steps")
plt.ylabel("Loss")
plt.title("Training / Validation Loss")
plt.legend()
plt.grid(True)
plt.show()
```

## Generative inference (free-text answers)

```python
def generate_answer(model, tokenizer, prompt):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=64,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[-1]:],
        skip_special_tokens=True
    ).strip()
```

## Evaluation

```python
def evaluate_model(model, tokenizer, dataset, max_samples=100):
    refs, preds = [], []

    for i, ex in enumerate(dataset):
        if i >= max_samples:
            break

        pred = generate_answer(model, tokenizer, ex["input"])
        refs.append(ex["output"].strip().lower())
        preds.append(pred.strip().lower())

    return refs, preds
```

## Baseline vs LoRA comparison

```python
def exact_match(refs, preds):
    refs = [r.strip().lower() for r in refs]
    preds = [p.strip().lower() for p in preds]
    return np.mean([r == p for r, p in zip(refs, preds)])

```

```python
# Baseline
refs_base, preds_base = evaluate_model(
    base_model,
    tokenizer,
    dataset_raw["test"]
)

# LoRA
refs_lora, preds_lora = evaluate_model(
    model,
    tokenizer,
    dataset_raw["test"]
)


results = pd.DataFrame([
    {
        "Model": "Baseline",
        "Exact Match": exact_match(refs_base, preds_base),
    },
    {
        "Model": "LoRA Fine-tuned",
        "Exact Match": exact_match(refs_lora, preds_lora),
    }
])

results
```
